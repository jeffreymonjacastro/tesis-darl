# Exploración de Distribuciones y Data Profiling - physionet

Este notebook realiza un análisis comparativo entre las particiones In-Distribution (Train) y Out-of-Distribution (OOD Test) del dataset `physionet` usando `data_profiling` (anteriormente `ydata-profiling`).


```python
from data_profiling import ProfileReport
from darl.data.get_dataset import load_dataset, find_project_root
```

```python
PROJECT_ROOT = find_project_root()

DATASET_NAME = "physionet"
OUTPUT_DIR = PROJECT_ROOT / "outputs" / "data_profiling" / DATASET_NAME
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
print(f"Project root: {PROJECT_ROOT}")
print(f"Output dir: {OUTPUT_DIR}")

dset = load_dataset(dataset_name=DATASET_NAME)
print("¡Cargado!")
```

```python
X_train, y_train, _, _ = dset.get_pandas(split="train")
X_validation, y_validation, _, _ = dset.get_pandas(split="validation")
X_test, y_test, _, _ = dset.get_pandas(split="id_test")
X_ood, y_ood, _, _ = dset.get_pandas(split="ood_test")

print(f"Train (ID): {X_train.shape}")
print(f"Validation (ID): {X_validation.shape}")
print(f"Test (ID) : {X_test.shape}")
print(f"OOD Test  : {X_ood.shape}")
```

```python
df_train = X_train.copy()
df_train["SepsisLabel"] = y_train

df_ood = X_ood.copy()
df_ood["SepsisLabel"] = y_ood

print("Generando reporte de Train (ID) para physionet...")
report_train = ProfileReport(
    df_train, title=f"{DATASET_NAME} - Train Split (ID)", minimal=True
)

print("Generando reporte de OOD Test para physionet...")
report_ood = ProfileReport(
    df_ood, title=f"{DATASET_NAME} - OOD Test Split", minimal=True
)

print("Guardando reportes individuales...")
report_train.to_file(OUTPUT_DIR / "report_train.html")
report_ood.to_file(OUTPUT_DIR / "report_ood.html")

print("Generando y guardando reporte comparativo (Train vs OOD)... ")
comparison_report = report_train.compare(report_ood)
comparison_report.to_file(OUTPUT_DIR / "report_comparison.html")
print("¡Listo! Reportes guardados en:", OUTPUT_DIR)
```
